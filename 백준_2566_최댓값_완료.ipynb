```python
row, col = 0, 0
max_num = 0
for x in range(9):
  arr = list(map(int,input().split()))
  max_t = max(arr)
  if max_t >= max_num:
    max_num = max_t
    row = x
    col = arr.index(max_t)+1
  print(row,col,end=" ")
  print(max_num)
```
