```python
# 이거 오답 나옴
a  = input()
alp = ['c=','c-','dz=','d-','lj','nj','s=','z=']
for x in alp:
  a = a.replace(x,"=")
print(len(a))
```

```python
# https://ooyoung.tistory.com/74
# [Python] 백준 알고리즘 온라인 저지 2941번 : 크로아티아 알파벳
# 이건 왜 되고 위에는 안되지.. 음?

croatia = ['c=', 'c-', 'dz=', 'd-', 'lj', 'nj', 's=', 'z=']
word = input()

for i in croatia :
    word = word.replace(i, '*')  # input 변수와 동일한 이름의 변수
print(len(word))
```

```python
# 이걸로 제출해서 정답 처리됨
alp = ['c=','c-','dz=','d-','lj','nj','s=','z=']
a  = input()
for x in alp:
  a = a.replace(x,"!") # 여기 "=" 넣으면 추후에 영향을 받을 것을 생각을 안했네 아...
print(len(a))
```
