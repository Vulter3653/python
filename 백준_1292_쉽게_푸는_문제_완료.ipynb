```python
a = 3
b = 7
# a,b = map(int,input().split())
print(a+b)
n = 1
num = ""
while b > len(num):
  # print(n*(n+1)//2)
  num += str(n)*n
  n += 1

print(n)
print(num)
print(list(num)[a-1:b])
print(sum(map(int,list(num)[a-1:b])))
```

```python
# 이거 오답
a,b = map(int,input().split())
n = 0
num = ""
while b > len(num):
  n += 1
  num += str(n)*n

print(sum(map(int,list(num)[a-1:b])))
```

```python
# https://sodehdt-ldkt.tistory.com/230
#1292
# 이건 왜 되는가

a, b = map(int, input().split())
sq = []
for i in range(1, b+1):
  for j in range(i):
    sq.append(i)
result = sum(sq[a-1:b])
print(result)
```
