```python
# 이거 통과함 얏호
a = []
max_lenth = 0

for x in range(5):
  plus = input()
  if len(plus) > max_lenth:
    max_lenth = len(plus)
  a.append(plus)

for x in range(5):
  if max_lenth > len(a[x]) :
    a[x] += "*"*(max_lenth-len(a[x]))

output = ""

for y in range(max_lenth):
  for x in range(5):
    output+=a[x][y]

print(output.replace("*",""))
```
